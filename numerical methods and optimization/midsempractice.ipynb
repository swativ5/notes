```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import linspace
import math
def f(x):
    return (5 * np.cos(2 * x)) - (2 * x * np.sin(2 * x))
t = linspace(1, 2, 1000)
plt.plot(t, f(t))
```

```python
y_values = f(t)

min_idx = np.argmin(y_values)
max_idx = np.argmax(y_values)

x_min = t[min_idx]
y_min = y_values[min_idx]
x_max = t[max_idx]
y_max = y_values[max_idx]

print(f"Minimum: f({x_min:.4f}) = {y_min:.4f}")
print(f"Maximum: f({x_max:.4f}) = {y_max:.4f}")

plt.figure(figsize=(10, 6))
plt.plot(t, y_values, 'b-', label='f(x)')
plt.plot(x_min, y_min, 'ro', markersize=10, label=f'Min at ({x_min:.4f}, {y_min:.4f})')
plt.plot(x_max, y_max, 'go', markersize=10, label=f'Max at ({x_max:.4f}, {y_max:.4f})')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def q51(x):
    return np.ones_like(x)

def q52(x):
    return -np.ones_like(x)

def q53(x):
    return 1 + 2*x

def q54(x):
    return -1 - 2*x

t1 = np.linspace(0,10,1000)
t2 = np.linspace(0,10,1000)
t3 = np.linspace(-10,0,1000)
t4 = np.linspace(-10,0,1000)

plt.figure(figsize=(10,6))

plt.plot(t1, q51(t1), label='y=1')
plt.plot(t2, q52(t2), label='y=-1')
plt.plot(t3, q53(t3), label='y=1+2x')
plt.plot(t4, q54(t4), label='y=-1-2x')


plt.legend()
plt.grid(True)
plt.show()
```

```python
from scipy.integrate import quad
def q8a(x):
    return np.sqrt(4 - (x ** 2))
integral8a, _ = quad(q8a, -2, 2)
t = np.linspace(-2, 2, 100)
plt.plot(t, q8a(t))
print(f"mean: {integral8a/4}")

def q8b(x):
    return (x ** 2) - 1
integral8b, _ = quad(q8b, 0, np.sqrt(3))
t = np.linspace(0, np.sqrt(3), 100)
plt.plot(t, q8b(t))
print(f"mean: {integral8b/np.sqrt(3):.4f}")

def q8c(x):
    return abs(x) - 1
integral8c, _ = quad(q8c, -1, 3)
t = np.linspace(-1, 3, 100)
plt.plot(t, q8c(t))
print(f"mean: {integral8c/4}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve


def q1(x):
    return np.exp(-x) - x
t = np.linspace(-10, 10, 100)
root = fsolve(q1, 0.3)[0]
plt.plot(t, q1(t))
print(root)

def bisectionx(f, a, b, tol):
    vals = list()
    while ((b - a) > tol):
        p = (a + b) / 2
        vals.append(p)
        if (f(p) == 0):
            return vals
        elif (f(a) * f(p) > 0):
            a = p
        else:
            b = p
    return vals
vals = bisectionx(q1, 0, 1, 0.000000000000000000001)
print(vals[-1])
```

```python
import math
def q2(x):
    return 30 * np.sin(x) + x ** 3 - 5
print(fsolve(q2, 0)[0])
bvals = bisectionx(q2, 0, 1.44, 0.001)
print(bvals[-1])

def dq2(x):
    return 30 * np.cos(x) + 3 * np.pow(x, 2) 

def newtonraphson(f, df, p0, tol):
    vals = [p0]
    p = p0 - f(p0)/df(p0)
    while (abs(p - p0) >= tol):
        p0 = p
        p = p0 - f(p0)/df(p0)
        vals.append(p)
    return vals
vals = newtonraphson(q2, dq2, 0, 0.001)
print(vals[-1])
print(len(bvals), len(vals))
print(newtonraphson(q2, dq2, 1.43, 0.00000001)[:4])
print(newtonraphson(q2, dq2, 1.44, 0.00000001)[:4])
print(newtonraphson(q2, dq2, 1.45, 0.00000001)[:4])
```

```python
def q6(x):
    return x - np.tan(x)
def dq6(x):
    return 1 - 1/np.cos(x)**2
t = np.linspace(-1, 2, 1000)
plt.plot(t, q6(t))
plt.xlim(-1, 4)
plt.ylim(-4, 4)
plt.grid('both')

def newtonraphson(f, df, p0, tol):
    vals = [p0]
    p = p0 - f(p0)/df(p0)
    while (abs(p - p0) <= tol):
        p0 = p
        if (abs(df(p0)) < 1e-10 or abs(f(p0)) < 1e-10):
            p0 = p0 + 0.001
        p = p0 - f(p0)/df(p0)
        vals.append(p)
    return vals

vals = newtonraphson(q6, dq6, 0.5, 1e-5)
print(fsolve(q6, 0.5)[0])
print(vals[-1])
vals = newtonraphson(q6, dq6, 102, 1e-5)
print(fsolve(q6, 102)[0])
print(vals[-1])
```
