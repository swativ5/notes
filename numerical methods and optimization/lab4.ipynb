P(x) = sum_i(f(x_i) * L_n,i(x))

where L_n,i(x) is the Lagrange basis polynomial defined as:L_n,i(x) = product_{j=0, j!=i}^{n} ((x - x_j) / (x_i - x_j))

|X_i|f(X_i)|
|---|---|
|0| 1|
|-1|2|
|1|3|

P(x) = 1 + x/2 + 3x^2/2

input: x
output: P(x)

```python
n = 3
x = [0, -1, 1]
y = [1, 2, 3]
```

```python
def lagrange(n, i, X):
    L = 1
    for j in range(0, n):
        if (i == j):
            continue
        L *= (X - x[j])/(x[i] - x[j])
    return L
```

```python
import numpy as np
def LN(x, X):
    n = len(x)
    Ln = []
    for i in range(n):
        Li = 1
        for j in range(n):
            if j != i:
                Li *= (X - x[j]) / (x[i] - x[j])
        Ln.append(Li)
    return np.array(Ln)

def Palt(X, n):
    Ln = LN(x, X)
    y_pred = 0
    for i in range(n):
        y_pred += (y[i] * Ln[i])
    return y_pred
```

```python
def P(X):
    y_pred = 0
    for i in range(0, n):
        y_pred += (y[i] * lagrange(n, i, X))
    return y_pred
```

```python
print("x\ty\tPalt\tP")
for i in range(0, n):
    print(f"{x[i]}\t{y[i]}\t{Palt(x[i])}\t{P(x[i])}")
```

```python
def taylor_func(x, a=1, n_terms=10):
    result = 0
    for n in range(n_terms):
        result += ((-1)**n) * ((x - a)**n) / (a**(n+1))
    return result   
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import linspace

x = [1, 1.3, 1.6, 1.9, 2.2]
y = [0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623]
n = 5

def LN(x, X, n):
    Ln = []
    for i in range(n):
        Li = np.ones_like(X)
        for j in range(n):
            if j != i:
                Li *= (X - x[j]) / (x[i] - x[j])
        Ln.append(Li)
    return np.array(Ln)

def Palt(X, n):
    Ln = LN(x, X, n)
    y_pred = 0
    for i in range(n):
        y_pred += (y[i] * Ln[i])
    return y_pred

t = linspace(1, 2.2, 1000)
colors = ['red', 'blue', 'green', 'orange', 'purple']

for i in range(1, 6):
    plt.plot(t, Palt(t, i), color=colors[i-1], label=f'P(t) with n={i}')

plt.scatter(x, y, color='black', s=50, zorder=5, label='Given Points')

plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
