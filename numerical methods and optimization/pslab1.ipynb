```python
import numpy as np
import matplotlib.pyplot as plt

# Sample function z = f(x, y)
x = np.linspace(-10, 10, 200)
y = np.linspace(-10, 10, 200)
X, Y = np.meshgrid(x, y)
Z = -(X**2 + Y**2)

# 1) plot in 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.show()


# 2) 2D contour
plt.figure(figsize=(6, 5))
c = plt.contour(X, Y, Z, levels=20, cmap="viridis")
plt.clabel(c, inline=True, fontsize=8)
plt.show()

# 3) 3D contour
fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection="3d")
ax.contour3D(X, Y, Z, levels=30, cmap="viridis")
plt.show()
```

```python
def contourz(X, Y, Z, h, tol):
    Xc, Yc, Zc = [], [], []
    for i in range(np.shape(X)[0]):
        for j in range(np.shape(Y)[0]):
            if abs(Z[i, j] - h) < tol:
                Xc.append(X[i, j])
                Yc.append(Y[i, j])
                Zc.append(Z[i, j])
    return [np.array(Xc), np.array(Yc), np.array(Zc)]                

h = -1
tol = 0.1
Xc, Yc, Zc = contourz(X, Y, Z, h, tol)
print(Xc)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# For 1D contour points, use scatter (or plot_trisurf), not plot_surface
ax.scatter(Xc, Yc, Zc)
ax.set_title(f"Points near z={h}")
plt.show()
```

```python
def f(x, y):
    return - x ** 2 - y ** 2
x = np.linspace(-5, 5, 1000)
y = x
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

Z1 = f(X*0, Y)
Z2 = f(X, Y*0)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.plot3D(X*0, Y, Z1, 'black')
ax.plot3D(X, Y*0, Z2, 'pink')
plt.show()
```
