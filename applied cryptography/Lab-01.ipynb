1. Basic Operations (GCD using Euclidean Algorithm)

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

print("GCD of 252 and 105 =", gcd(252, 105))
```

2. Modular Arithmetic

```python
def modular_operations(a, b, n):
    add = (a + b) % n
    multiply = (a * b) % n
    return add, multiply


add, multiply = modular_operations(15, 37, 13)
print("(15 + 37) mod 13 =", add)
print("(15 * 37) mod 13 =", multiply)
```

3. Fast Modular Exponentiation

```python
def mod_exp(base, exp, mod):
    result = 1
    base %= mod

    while exp > 0:
        if exp % 2 == 1:
            result = (result * base) % mod
        base = (base * base) % mod
        exp //= 2

    return result

print("3^50 mod 7 =", mod_exp(3, 50, 7))
print("2^100 mod 13 =", mod_exp(2, 100, 13))
print("5^200 mod 11 =", mod_exp(5, 200, 11))
```
